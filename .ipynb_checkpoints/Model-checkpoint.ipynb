```python
import pandas as pd
```

```python
df=pd.read_csv('Salary_Data[1].csv')
```

```python
df
```

```python
df.isnull().sum()
```

```python
df['Age']=df['Age'].fillna(df['Age'].mean())
df["Years of Experience"]=df["Years of Experience"].fillna(df["Years of Experience"].mean())
df["Salary"]=df["Salary"].fillna(df["Salary"].mean())
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
df['Education Level'] = df['Education Level'].fillna(df['Education Level'].mode()[0])
df['Job Title'] = df['Job Title'].fillna(df['Job Title'].mode()[0])
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np
```

```python
df['Years of Experience']=df['Years of Experience'].astype(int)
df["Age"] = df["Age"].astype(int)
df["Salary"]=df["Salary"].astype(int)
```

```python
X=df[['Years of Experience']]
Y=df['Salary']
model=LinearRegression()
model.fit(X,Y)
```

```python
print('Slope (m):', model.coef_) 
print('Intercept (c):', model.intercept_) 
```

```python
print('Prediction for 10 years of experience',model.predict([[10]]))
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
model.fit(X_train,Y_train)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
print("MAE :", mean_absolute_error(Y_test, y_pred))
print("MSE :", mean_squared_error(Y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(Y_test, y_pred)))
print("R² Score:", r2_score(Y_test, y_pred)) #proportion of variance
```

Gradient Descent

```python
df=pd.read_csv('Salary_Data[1].csv')
```

```python
df
```

```python
X=df['Years of Experience']
Y=df['Salary']
n=len(X)
m, c = 0.0, 0.0       
lr = 0.00001          
epochs = 20000

loss_history = []

for i in range(epochs):
    Y_pred = m * X + c
    error  = Y_pred - Y

    
    loss = (1/n) * np.sum(error ** 2)
    loss_history.append(loss)

    dm = (2/n) * np.sum(error * X)
    dc = (2/n) * np.sum(error)

    
    m = m - lr * dm
    c = c - lr * dc

    if i % 4000 == 0:
        print(f'epoch {i}: loss={loss:.2f}  m={m:.3f}  c={c:.3f}')

print('Final parameters -> m:', m, ' c:', c)
```

```python
import matplotlib.pyplot as plt


plt.plot(range(epochs), loss_history, color='blue')
plt.xlabel('Epochs')
plt.ylabel('Loss (MSE)')
plt.title('Loss vs Epochs (Convergence)')
plt.grid(True)
plt.show()
```

```python
print("First 10 losses:", loss_history[:10])
print("\nLast 10 losses:", loss_history[-10:])
```

Multiple Linear Regression

```python
import pandas as pd
import numpy as np
X = df[['Years of Experience', 'Age']].values.astype(float)  
Y = df['Salary'].values.astype(float)                        
n, k = X.shape


m = np.zeros(k)  
c = 0.0          
lr = 0.00001
epochs = 20000

loss_history = []

for i in range(epochs):
    Y_pred = np.dot(X, m) + c
    error  = Y_pred - Y

    
    loss = (1/n) * np.sum(error ** 2)
    loss_history.append(loss)

    dm = (2/n) * np.dot(X.T, error)   
    dc = (2/n) * np.sum(error)

    m = m - lr * dm
    c = c - lr * dc

    if i % 4000 == 0:
        print(f'epoch {i}: loss={loss:.2f}  m={m}  c={c:.3f}')

print('Final parameters -> m:', m, ' c:', c)
```
